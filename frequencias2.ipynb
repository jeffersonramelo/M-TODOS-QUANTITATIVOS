```python
import pandas as pd

#lendo os dados
dados = pd.read_csv("/content/Exercício 02_dados_prazo_recebimento.csv")
dados
```

```python
#isolar a coluna que formará a tabela de frequências
x = dados["Prazo_recebimento"]
len(x)

#ordenação em ROL
y = x.sort_values()
y
```

```python
# Frequência simples absoluta
fs = y.value_counts().sort_index()

fs

# Frequência simples relativa (%)
fsr = (fs / len(x)) * 100

fsr

#Frequência acumulada
fa = fs.cumsum()

fa

#Frequência relativa acumulada (%)
far = (fa / len(x)) * 100

far
```

```python
#juntar todas as frequências

tabela = pd.DataFrame({"fsimples": fs, "fsimples%": fsr, "facumulada": fa, "facumulada%": far})
tabela
#
```

```python
#organizada em classes de frequências

import numpy as np

z = np.sqrt(len(x))
z
```

```python
max = x.max()
max
min = x.min()
min

h = (max - min)/z
h
```

```python
limites = (4,8,12,16,20,24,28,33)

import pandas as pd

classes = pd.cut(
    x,
    bins=limites,
    right=False
)

classes
```

```python
# Frequência simples absoluta por classe
fs_classes = classes.value_counts().sort_index()
fs_classes
```

```python
# Frequência simples relativa (%) por classe
fsr_classes = (fs_classes / len(x)) * 100
fsr_classes
```

```python
# Frequência acumulada por classe
fa_classes = fs_classes.cumsum()
fa_classes
```

```python
# Frequência relativa acumulada (%) por classe
far_classes = (fa_classes / len(x)) * 100
far_classes
```

```python
# Juntar todas as frequências por classes em uma tabela
tabela_classes = pd.DataFrame({"fsimples_classe": fs_classes, "fsimples%_classe": fsr_classes, "facumulada_classe": fa_classes, "facumulada%_classe": far_classes})
tabela_classes
```
