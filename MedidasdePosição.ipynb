```python
# patinetes elétricos percorrem 20km, com as seguintes velocidades:
import pandas as pd
import statistics as st
vel=(6,8,12,30)
# como calcular a média? a média aritmética é adequada?
st.mean(vel)
```

```python
# Calculate the harmonic mean
harmonic_mean = st.harmonic_mean(vel)
print("Harmonic Mean:", round(harmonic_mean, 2))
```
